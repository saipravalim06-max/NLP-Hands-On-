```python
def gender_features(word):
    return {'last_letter':word[-1]}

gender_features('Obama')
```

```python
import nltk

nltk.download('names')
```

```python
from nltk.corpus import names

names.words()
print(len(names.words()))
```

```python
labeled_names = (
    [(name, 'male') for name in names.words('male.txt')] +
    [(name, 'female') for name in names.words('female.txt')]
)
```

```python
import random

random.shuffle(labeled_names)

featuresets = [
    (gender_features(n), gender)
    for (n, gender) in labeled_names
]

# Try printing feature sets
print(featuresets[:10])

train_set, test_set = featuresets[5000:], featuresets[:2000]
```

```python
import nltk

classifier = nltk.NaiveBayesClassifier.train(train_set)

classifier.classify(gender_features('David'))
print(nltk.classify.accuracy(classifier, test_set))

classifier.show_most_informative_features(10)
```

```python
classifier = nltk.NaiveBayesClassifier.train(train_set)

classifier.classify(gender_features('Pravali'))
print(nltk.classify.accuracy(classifier, test_set))

classifier.show_most_informative_features(10)
```

```python
#Experiment 2-Text vectorization using CountVectorizer

from sklearn.feature_extraction.text import CountVectorizer

vect = CountVectorizer(binary=True)

corpus = [
    "Tessaract is good optical character recognition engine",
    "optical character recognition is significant"
]

vect.fit(corpus)

# Stores the vocabulary
vocab = vect.vocabulary_

for key in sorted(vocab.keys()):
    print("{}:{}".format(key, vocab[key]))
```

```python
print(
    vect.transform(
        ["This is a good optical illusion"]
    ).toarray()
)
```

```python
#Experiment 3-Finding Similarity Between the Documents
from sklearn.metrics.pairwise import cosine_similarity

similarity = cosine_similarity(
    vect.transform(
        ["Google Cloud Vision is a character recognition engine"]
    ).toarray(),
    vect.transform(
        ["OCR is an optical character recognition engine"]
    ).toarray()
)

print(similarity)
```
